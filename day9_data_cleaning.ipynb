# Day 9 – Data Cleaning & Preparation

## Cleaning Summary

The Superstore dataset was inspected and cleaned to improve data quality and prepare it for further analysis.

### 1. Missing Values
- 11 missing values were found in the `Postal Code` column.
- The missing values were replaced with `0`.
- `0` represents an unknown/missing postal code.
- Other columns had no missing values.

### 2. Duplicate Records
- The dataset was checked for duplicate rows.
- No duplicate records were found.
- Therefore, no rows were removed.

### 3. Date Format
- `Order Date` and `Ship Date` were initially stored as `object` (text).
- Both columns were converted to `datetime64[ns]`.
- `dayfirst=True` was used because the dates were in DD/MM/YYYY format.

### 4. Date Consistency
- The dataset was checked for records where the Ship Date occurred before the Order Date.
- No such records were found.

### 5. Categorical Data
- Categorical columns were checked for inconsistent values, capitalization, and spelling.
- No obvious inconsistencies were found.
- Therefore, no changes were made.

### 6. Extra Spaces
- Text columns were checked for leading and trailing spaces.
- No extra spaces were found.

### 7. Numeric Data
- Numeric columns were inspected using descriptive statistics.
- No negative or zero Sales values were found.
- No additional numeric cleaning was required.

## Final Data Quality

- Rows: 9,800

- Duplicate rows: 0
- Negative Sales: 0
- Zero Sales: 0
