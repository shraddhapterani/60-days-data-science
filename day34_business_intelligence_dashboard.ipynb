## ABTalks Data Science Challenge

### Objective
Build an interactive customer analytics dashboard
for business decision-making.

### Business Problem
Executives need a clear view of customer performance,
churn risks, and customer segments to make informed decisions.

### Dashboard Features
1. Customer KPI cards
2. Customer segmentation insights
3. Churn analysis
4. Interactive charts and filters
5. Business recommendations

### Tools
Python, Pandas, Matplotlib, Plotly, Streamlit

```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Create customer data
n = 300

df = pd.DataFrame({
    "Customer_ID": [f"C{i:03d}" for i in range(1, n + 1)],
    "Age": np.random.randint(18, 65, n),
    "Monthly_Spend": np.random.randint(500, 15000, n),
    "Tenure_Months": np.random.randint(1, 60, n),
    "Support_Calls": np.random.poisson(2, n),
    "Churn": np.random.choice(
        ["Yes", "No"], n, p=[0.25, 0.75]
    ),
    "Segment": np.random.choice(
        ["Regular", "Premium", "Budget"], n
    )
})

print("Dataset created successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
import plotly.graph_objects as go
from IPython.display import display, HTML

# Calculate business KPIs
total_customers = len(df)

avg_spend = df["Monthly_Spend"].mean()

churn_rate = (df["Churn"] == "Yes").mean() * 100

avg_tenure = df["Tenure_Months"].mean()

# Display KPI cards
kpis = [
    ("Total Customers", f"{total_customers:,}"),
    ("Avg Monthly Spend", f"₹{avg_spend:,.0f}"),
    ("Churn Rate", f"{churn_rate:.1f}%"),
    ("Avg Tenure", f"{avg_tenure:.1f} months")
]

cards = ""

for title, value in kpis:
    cards += f"""
    <div style="
        background:#f0f4ff;
        padding:20px;
        border-radius:12px;
        text-align:center;
        flex:1;
        min-width:150px;
        margin:8px;">
        <h4 style="color:#555;">{title}</h4>
        <h2 style="color:#4338ca;">{value}</h2>
    </div>
    """

display(HTML(
    f'<div style="display:flex;flex-wrap:wrap;">{cards}</div>'
))
```

```python
import plotly.express as px

segment_counts = df["Segment"].value_counts().reset_index()
segment_counts.columns = ["Segment", "Customers"]

fig = px.pie(
    segment_counts,
    names="Segment",
    values="Customers",
    title="Customer Segmentation Distribution",
    hole=0.4
)

fig.show()
```

```python
churn_counts = df["Churn"].value_counts().reset_index()
churn_counts.columns = ["Churn", "Customers"]

fig = px.bar(
    churn_counts,
    x="Churn",
    y="Customers",
    color="Churn",
    title="Customer Churn Analysis",
    text="Customers"
)

fig.show()
```

```python
fig = px.scatter(
    df,
    x="Tenure_Months",
    y="Monthly_Spend",
    color="Churn",
    size="Support_Calls",
    hover_name="Customer_ID",
    title="Customer Spending vs Tenure",
    labels={
        "Tenure_Months": "Tenure (Months)",
        "Monthly_Spend": "Monthly Spend (₹)",
        "Churn": "Churn Status"
    }
)

fig.show()
```

```python
segment_churn = (
    df.groupby(["Segment", "Churn"])
    .size()
    .reset_index(name="Customers")
)

fig = px.bar(
    segment_churn,
    x="Segment",
    y="Customers",
    color="Churn",
    barmode="group",
    title="Customer Churn by Segment",
    text="Customers"
)

fig.show()
```

```python
import plotly.express as px

fig = px.bar(
    df,
    x="Segment",
    y="Monthly_Spend",
    color="Churn",
    title="Monthly Spending by Segment and Churn",
    barmode="group",
    hover_data=["Customer_ID", "Tenure_Months", "Support_Calls"]
)

fig.show()
```

```python
# Export customer segmentation report
segment_report = (
    df.groupby("Segment")
    .agg(
        Total_Customers=("Customer_ID", "count"),
        Average_Spend=("Monthly_Spend", "mean"),
        Churned_Customers=("Churn", lambda x: (x == "Yes").sum())
    )
    .reset_index()
)

segment_report.to_csv(
    "day34_segment_dashboard_report.csv",
    index=False
)

# Export churn report
churn_report = (
    df.groupby("Churn")
    .agg(
        Total_Customers=("Customer_ID", "count"),
        Average_Spend=("Monthly_Spend", "mean"),
        Average_Tenure=("Tenure_Months", "mean")
    )
    .reset_index()
)

churn_report.to_csv(
    "day34_churn_dashboard_report.csv",
    index=False
)

print("Day 34 reports exported successfully!")
```
