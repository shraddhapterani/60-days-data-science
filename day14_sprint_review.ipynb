# Day 13 - Overfitting and Regularization

## ABTalks 60-Day Data Science Challenge

Comparing Linear Regression, Ridge Regression, and Lasso Regression to understand overfitting, regularization, and model generalization.

```python
import pandas as pd

df = pd.read_csv("superstore_sales_cleaned.csv")

df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

df['Shipping Days'] = (
    df['Ship Date'] - df['Order Date']
).dt.days

df['Order Year'] = df['Order Date'].dt.year

df.head()
```

```python
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

df['Shipping Days'] = (
    df['Ship Date'] - df['Order Date']
).dt.days

df['Order Year'] = df['Order Date'].dt.year

df.head()
```

```python
features = [
    'Shipping Days',
    'Order Year',
    'Category',
    'Sub-Category',
    'Region',
    'Ship Mode',
    'Segment'
]

X = df[features]
y = df['Sales']

print("Input features:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

```python
X_encoded = pd.get_dummies(
    X,
    columns=[
        'Category',
        'Sub-Category',
        'Region',
        'Ship Mode',
        'Segment'
    ],
    dtype=int
)

print("Before encoding:", X.shape)
print("After encoding:", X_encoded.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

baseline_model = LinearRegression()

baseline_model.fit(X_train, y_train)

print("Baseline Linear Regression trained successfully!")
```

```python
from sklearn.metrics import mean_absolute_error, r2_score

# Predictions
baseline_train_pred = baseline_model.predict(X_train)
baseline_test_pred = baseline_model.predict(X_test)

# MAE
baseline_train_mae = mean_absolute_error(y_train, baseline_train_pred)
baseline_test_mae = mean_absolute_error(y_test, baseline_test_pred)

# R²
baseline_train_r2 = r2_score(y_train, baseline_train_pred)
baseline_test_r2 = r2_score(y_test, baseline_test_pred)

print("Baseline Model Performance")
print("---------------------------")
print("Train MAE:", baseline_train_mae)
print("Test MAE:", baseline_test_mae)
print("Train R²:", baseline_train_r2)
print("Test R²:", baseline_test_r2)
```

```python
from sklearn.linear_model import Ridge

ridge_model = Ridge(alpha=1.0)

ridge_model.fit(X_train, y_train)

print("Ridge Regression trained successfully!")
```

```python
ridge_train_pred = ridge_model.predict(X_train)
ridge_test_pred = ridge_model.predict(X_test)

ridge_train_mae = mean_absolute_error(y_train, ridge_train_pred)
ridge_test_mae = mean_absolute_error(y_test, ridge_test_pred)

ridge_train_r2 = r2_score(y_train, ridge_train_pred)
ridge_test_r2 = r2_score(y_test, ridge_test_pred)

print("Ridge Regression Performance")
print("----------------------------")
print("Train MAE:", ridge_train_mae)
print("Test MAE:", ridge_test_mae)
print("Train R²:", ridge_train_r2)
print("Test R²:", ridge_test_r2)
```

```python
from sklearn.linear_model import Lasso

lasso_model = Lasso(alpha=1.0)

lasso_model.fit(X_train, y_train)

print("Lasso Regression trained successfully!")
```

```python
lasso_train_pred = lasso_model.predict(X_train)
lasso_test_pred = lasso_model.predict(X_test)

lasso_train_mae = mean_absolute_error(y_train, lasso_train_pred)
lasso_test_mae = mean_absolute_error(y_test, lasso_test_pred)

lasso_train_r2 = r2_score(y_train, lasso_train_pred)
lasso_test_r2 = r2_score(y_test, lasso_test_pred)

print("Lasso Regression Performance")
print("----------------------------")
print("Train MAE:", lasso_train_mae)
print("Test MAE:", lasso_test_mae)
print("Train R²:", lasso_train_r2)
print("Test R²:", lasso_test_r2)
```

```python
comparison = pd.DataFrame({
    'Model': ['Linear Regression', 'Ridge', 'Lasso'],
    'Train MAE': [
        baseline_train_mae,
        ridge_train_mae,
        lasso_train_mae
    ],
    'Test MAE': [
        baseline_test_mae,
        ridge_test_mae,
        lasso_test_mae
    ],
    'Train R²': [
        baseline_train_r2,
        ridge_train_r2,
        lasso_train_r2
    ],
    'Test R²': [
        baseline_test_r2,
        ridge_test_r2,
        lasso_test_r2
    ]
})

comparison
```

```python
comparison['R² Gap'] = (
    comparison['Train R²'] - comparison['Test R²']
)

comparison
```

```python
import numpy as np

lasso_coefficients = pd.Series(
    lasso_model.coef_,
    index=X_train.columns
)

near_zero = np.sum(np.isclose(lasso_coefficients, 0))

print("Total coefficients:", len(lasso_coefficients))
print("Coefficients near zero:", near_zero)
```

```python
zero_features = lasso_coefficients[
    np.isclose(lasso_coefficients, 0)
]

print("Features with coefficients near zero:")
print(zero_features)
```

```python
best_mae_model = comparison.loc[
    comparison['Test MAE'].idxmin(), 'Model'
]

best_r2_model = comparison.loc[
    comparison['Test R²'].idxmax(), 'Model'
]

print("Best model by Test MAE:", best_mae_model)
print("Best model by Test R²:", best_r2_model)
```

## Day 13 Conclusion

### Model Comparison

Three regression models were compared:

- Linear Regression
- Ridge Regression
- Lasso Regression

Linear Regression achieved the best test performance with:
- Test MAE: 238.77
- Test R²: 0.179

Ridge Regression produced almost identical results to Linear Regression, showing that regularization with alpha=1.0 had only a small effect.

Lasso Regression reduced 14 out of 33 coefficients to approximately zero, demonstrating feature selection. However, its test performance was slightly lower.

### Overfitting Analysis

The train-test R² gaps were small:

- Linear Regression: 0.026
- Ridge: 0.026
- Lasso: 0.031

Therefore, there was no strong evidence of overfitting in this experiment.

### Key Learning

High training performance alone is not enough. Models should also perform well on unseen data.

Regularization can control model complexity and reduce the influence of unnecessary features, but it does not always improve test performance. The effect depends on the dataset and model.

```python
import pandas as pd

df = pd.read_csv("superstore_sales_cleaned.csv")

df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

df['Shipping Days'] = (
    df['Ship Date'] - df['Order Date']
).dt.days

df['Order Year'] = df['Order Date'].dt.year

print(df.shape)
```

```python
features_without_subcategory = [
    'Shipping Days',
    'Order Year',
    'Category',
    'Region',
    'Ship Mode',
    'Segment'
]

X_reduced = df[features_without_subcategory]

X_reduced_encoded = pd.get_dummies(
    X_reduced,
    columns=[
        'Category',
        'Region',
        'Ship Mode',
        'Segment'
    ],
    dtype=int
)

print("Features after removal:", X_reduced_encoded.shape[1])
```

```python
from sklearn.model_selection import train_test_split

y = df['Sales']

X_train_reduced, X_test_reduced, y_train_reduced, y_test_reduced = train_test_split(
    X_reduced_encoded,
    y,
    test_size=0.2,
    random_state=42
)

print("Training data:", X_train_reduced.shape)
print("Testing data:", X_test_reduced.shape)
```

```python
from sklearn.linear_model import LinearRegression

reduced_model = LinearRegression()

reduced_model.fit(X_train_reduced, y_train_reduced)

print("Reduced Linear Regression trained successfully!")
```

```python
from sklearn.metrics import mean_absolute_error, r2_score

reduced_train_pred = reduced_model.predict(X_train_reduced)
reduced_test_pred = reduced_model.predict(X_test_reduced)

reduced_train_mae = mean_absolute_error(y_train_reduced, reduced_train_pred)
reduced_test_mae = mean_absolute_error(y_test_reduced, reduced_test_pred)

reduced_train_r2 = r2_score(y_train_reduced, reduced_train_pred)
reduced_test_r2 = r2_score(y_test_reduced, reduced_test_pred)

print("Model WITHOUT Sub-Category")
print("--------------------------")
print("Train MAE:", reduced_train_mae)
print("Test MAE:", reduced_test_mae)
print("Train R²:", reduced_train_r2)
print("Test R²:", reduced_test_r2)
```
