```python
import pandas as pd

df = pd.read_csv("superstore_sales_cleaned.csv")

df.head()
```

```python
numeric_features = df.select_dtypes(include=['int64', 'float64']).columns
categorical_features = df.select_dtypes(include=['object']).columns

print("Numerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days
```

```python
df['Sales Per Quantity'] = df['Sales'] / df['Quantity']
```

```python
df['Order Year'] = df['Order Date'].dt.year
```

```python
df[['Order Date', 'Shipping Days', 'Order Year']].head()
```

```python
df['Category'].value_counts()
```

```python
df = pd.get_dummies(df, columns=['Category'], dtype=int)
```

```python
df.head()
```

```python
categorical_to_encode = [
    'Ship Mode',
    'Segment',
    'Region',
    'Sub-Category'
]

df = pd.get_dummies(
    df,
    columns=categorical_to_encode,
    dtype=int
)
```

```python
df.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df[['Sales', 'Shipping Days']] = scaler.fit_transform(
    df[['Sales', 'Shipping Days']]
)
```
