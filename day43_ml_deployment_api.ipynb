```python
# Day 43 - ML Model Deployment API

import pandas as pd
import numpy as np

# Load Day 41 customer risk results
customer_risk = pd.read_csv("day41_customer_risk_final.csv")

print("Day 41 customer risk data loaded successfully.")
print("Shape:", customer_risk.shape)

customer_risk.head()
```

```python
# Day 43 - Step 2: Select Prediction Features

features = [
    "Recency_Days",
    "Purchase_Frequency",
    "Estimated_CLV"
]

target = "Risk_Level"

X = customer_risk[features]
y = customer_risk[target]

print("Prediction Features:")
print(X.head())

print("\nTarget Classes:")
print(y.value_counts())
```

```python
# Day 43 - Step 3: Train Risk Prediction Model

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Create the model
risk_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train the model
risk_model.fit(X_train, y_train)

# Make predictions
y_pred = risk_model.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)

print("Risk Prediction Model trained successfully.")
print(f"Accuracy: {accuracy:.2%}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Day 43 - Step 4: Save the Trained Model

import joblib

model_filename = "day43_customer_risk_model.pkl"

joblib.dump(risk_model, model_filename)

print("Model saved successfully.")
print(f"File: {model_filename}")
```
