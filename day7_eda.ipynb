```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("superstore_sales_cleaned.csv")
print(df.shape)
```

```python
print(df['Sales'].describe())
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Sales'], bins=50, kde=True)
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Category', y='Sales', data=df, estimator=sum, errorbar=None)
plt.title("Total Sales by Category")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Region', y='Sales', data=df, estimator=sum, errorbar=None)
plt.title("Total Sales by Region")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Sales'])
plt.title("Sales Outliers")
plt.show()
```

```python
plt.figure(figsize=(10,5))
monthly_sales = df.groupby('Order Month')['Sales'].sum()
monthly_sales.plot(kind='line', marker='o')
plt.title("Sales Trend by Month")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.show()
```

## Key Insights from EDA

1. **Technology** is the top-performing category by total sales (~₹850K), narrowly ahead of Furniture and Office Supplies (~₹720-730K each) — a fairly balanced spread across categories.

2. Sales are unevenly distributed across regions — the top-performing region generates roughly **double** the sales of the lowest-performing region, suggesting an opportunity to investigate what's driving regional performance gaps.

3. The Sales distribution is heavily **right-skewed**: most orders are small (median ₹54), but a handful of large orders (up to ₹22,638) pull the average up to ₹230 — a classic sign of long-tail business, common in retail.

4. The boxplot confirms numerous high-value outliers beyond the typical range, indicating a mix of everyday small purchases alongside occasional bulk or high-ticket orders.

5. Sales show a clear **seasonal pattern**, rising sharply from month 8 (August) through month 11 (November), likely driven by holiday season shopping — a useful signal for inventory and marketing planning.
