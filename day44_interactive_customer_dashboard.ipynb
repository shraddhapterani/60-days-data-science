```python
import pandas as pd
import numpy as np

customer_data = pd.read_csv("day41_customer_risk_final.csv")

print("Customer risk data loaded successfully.")
print("Shape:", customer_data.shape)

customer_data.head()
```

```python
print("Data Types:")
print(customer_data.dtypes)

print("\nMissing Values:")
print(customer_data.isnull().sum())
```

```python
customer_data["Last_Order_Date"] = pd.to_datetime(
    customer_data["Last_Order_Date"]
)

customer_data["First_Order_Date"] = pd.to_datetime(
    customer_data["First_Order_Date"]
)

print("Date columns converted successfully.")
print(customer_data[["Last_Order_Date", "First_Order_Date"]].dtypes)
```

```python
total_customers = customer_data["Customer ID"].nunique()
total_sales = customer_data["Total_Sales"].sum()
total_orders = customer_data["Total_Orders"].sum()
average_order_value = customer_data["Average_Order_Value"].mean()
high_risk_customers = (customer_data["Risk_Level"] == "High Risk").sum()
medium_risk_customers = (customer_data["Risk_Level"] == "Medium Risk").sum()
low_risk_customers = (customer_data["Risk_Level"] == "Low Risk").sum()

print("Customer Intelligence KPIs")
print("--------------------------")
print("Total Customers:", total_customers)
print("Total Sales:", round(total_sales, 2))
print("Total Orders:", total_orders)
print("Average Order Value:", round(average_order_value, 2))
print("High Risk Customers:", high_risk_customers)
print("Medium Risk Customers:", medium_risk_customers)
print("Low Risk Customers:", low_risk_customers)
```

```python
risk_distribution = (
    customer_data["Risk_Level"]
    .value_counts()
    .reset_index()
)

risk_distribution.columns = ["Risk_Level", "Customer_Count"]

print("Risk Distribution:")
print(risk_distribution)
```

```python
retention_distribution = (
    customer_data["Retention_Strategy"]
    .value_counts()
    .reset_index()
)

retention_distribution.columns = [
    "Retention_Strategy",
    "Customer_Count"
]

print("Retention Strategy Distribution:")
print(retention_distribution)
```
