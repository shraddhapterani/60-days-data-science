```python
import pandas as pd

df = pd.read_csv("superstore_sales.csv")
print("Original shape:", df.shape)
```

```python
print("Missing values per column:")
print(df.isnull().sum())
```

```python
# Drop rows with any missing values (since this dataset is usually clean, this is a safety step)
df = df.dropna()
print("Shape after removing missing values:", df.shape)
```

```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)

df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

```python
print("Data types before fixing:")
print(df.dtypes)

# Convert Order Date and Ship Date to proper datetime format
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d/%m/%Y')

print("\nData types after fixing:")
print(df.dtypes)
```

```python
# Feature 1: Shipping Duration (days between order and ship date)
df['Shipping Duration'] = (df['Ship Date'] - df['Order Date']).dt.days

# Feature 2: Order Month (extract just the month, useful for trend analysis)
df['Order Month'] = df['Order Date'].dt.month

df[['Order Date', 'Ship Date', 'Shipping Duration', 'Order Month']].head()
```

```python
df.to_csv("superstore_sales_cleaned.csv", index=False)
print("Cleaned dataset saved!")
```
