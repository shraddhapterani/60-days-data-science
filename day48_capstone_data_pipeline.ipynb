```python
import pandas as pd
from pathlib import Path


def preprocess_sales(input_path):
    """Clean sales data and create reusable analytical features."""

    # Load raw data
    data = pd.read_csv(input_path)

    # Remove exact duplicate rows
    data = data.drop_duplicates().copy()

    # Convert dates using the dataset's day/month/year format
    data["Order Date"] = pd.to_datetime(
        data["Order Date"],
        format="%d/%m/%Y",
        errors="coerce"
    )

    data["Ship Date"] = pd.to_datetime(
        data["Ship Date"],
        format="%d/%m/%Y",
        errors="coerce"
    )

    # Validate required dates
    if data[["Order Date", "Ship Date"]].isna().any().any():
        raise ValueError("Invalid or missing order/ship dates found.")

    # Create time-based features
    data["Order Year"] = data["Order Date"].dt.year
    data["Order Month"] = data["Order Date"].dt.month

    data["Shipping Days"] = (
        data["Ship Date"] - data["Order Date"]
    ).dt.days

    # Validate shipping duration
    if (data["Shipping Days"] < 0).any():
        raise ValueError("Negative shipping durations found.")

    # Preserve missing postal codes safely
    data["Postal Code"] = data["Postal Code"].astype("Int64")

    return data


# Run the reusable pipeline
cleaned_df = preprocess_sales("superstore_sales.csv")

# Save the cleaned dataset
output_path = Path("day48_superstore_sales_cleaned.csv")
cleaned_df.to_csv(output_path, index=False)

print("Preprocessing completed successfully!")
print("Output file:", output_path)
print("Dataset shape:", cleaned_df.shape)
print("Duplicate rows:", cleaned_df.duplicated().sum())
print("Missing postal codes:", cleaned_df["Postal Code"].isna().sum())
print("Missing dates:", cleaned_df[["Order Date", "Ship Date"]].isna().sum().sum())
```
