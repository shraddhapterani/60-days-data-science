```python
import pandas as pd
import os

# Check your project folder
print("Current folder:", os.getcwd())

# Week 5 datasets
files = [
    "day31_customer_persona_report.csv",
    "day32_recommendation_outputs.csv",
    "day33_anomaly_detection_results.csv",
    "day34_segment_dashboard_report.csv",
    "day34_churn_dashboard_report.csv"
]

# Check which files are available
for file in files:
    if os.path.exists(file):
        print("Found:", file)
    else:
        print("Missing:", file)
```

```python
import pandas as pd

# Load the five datasets
persona_df = pd.read_csv("day31_customer_persona_report.csv")
recommendation_df = pd.read_csv("day32_recommendation_outputs.csv")
anomaly_df = pd.read_csv("day33_anomaly_detection_results.csv")
segment_df = pd.read_csv("day34_segment_dashboard_report.csv")
churn_df = pd.read_csv("day34_churn_dashboard_report.csv")

# Display dataset dimensions
print("Customer Personas:", persona_df.shape)
print("Recommendations:", recommendation_df.shape)
print("Anomaly Detection:", anomaly_df.shape)
print("Segment Dashboard:", segment_df.shape)
print("Churn Dashboard:", churn_df.shape)
```

```python
datasets = {
    "Customer Personas": persona_df,
    "Recommendations": recommendation_df,
    "Anomaly Detection": anomaly_df,
    "Segment Dashboard": segment_df,
    "Churn Dashboard": churn_df
}

for name, df in datasets.items():
    print(f"\n--- {name} ---")
    print("Columns:", df.columns.tolist())
    display(df.head(3))
```

```python
# Create a unified summary of all four analytics modules

total_customers = churn_df["Total_Customers"].sum()
total_churned = churn_df.loc[
    churn_df["Churn"] == "Yes", "Total_Customers"
].sum()

total_anomalies = (anomaly_df["Anomaly"] == -1).sum()

print("===== DAY 35: INTEGRATED ANALYTICS SYSTEM =====")

print("\n1. CUSTOMER SEGMENTATION")
print("Number of personas:", len(persona_df))
print("Customer segments:", len(segment_df))

print("\n2. RECOMMENDATION SYSTEM")
print("Recommendations generated:", len(recommendation_df))

print("\n3. ANOMALY DETECTION")
print("Transactions analyzed:", len(anomaly_df))
print("Anomalies detected:", total_anomalies)

print("\n4. CUSTOMER CHURN")
print("Total customers:", total_customers)
print("Churned customers:", total_churned)
print("Churn rate:", round(total_churned / total_customers * 100, 2), "%")
```
